```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import CountVectorizer
from joblib import dump
import pandas as pd

# Example dataset (replace this with your actual dataset)
data = pd.read_csv('projects_dataset.csv')

# Extract features and labels
X = data['description'] + " " + data['synopsis']  # Combine text features
y = data['review_status']

# Vectorize the text features
vectorizer = CountVectorizer()
X_vectorized = vectorizer.fit_transform(X)

# Train Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_vectorized, y)

# Save the model and vectorizer
dump(model, 'models/random_forest_model.joblib')
dump(vectorizer, 'models/vectorizer.joblib')

print("Model and vectorizer saved!")
```

```python
import os
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import CountVectorizer
from joblib import dump
import pandas as pd

# Ensure the models directory exists
if not os.path.exists('models'):
    os.makedirs('models')

# Load dataset
try:
    data = pd.read_csv('projects_dataset.csv')
except FileNotFoundError:
    print("Error: 'projects_dataset.csv' not found. Please ensure the file exists.")
    exit()

# Check for required columns
required_columns = ['description', 'synopsis', 'review_status']
if not all(column in data.columns for column in required_columns):
    print(f"Error: Dataset is missing one or more required columns: {required_columns}")
    exit()

# Extract features and labels
X = data['description'] + " " + data['synopsis']  # Combine text features
y = data['review_status']

# Vectorize the text features
vectorizer = CountVectorizer()
X_vectorized = vectorizer.fit_transform(X)

# Train Random Forest model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_vectorized, y)

# Save the model and vectorizer
dump(model, 'models/random_forest_model.joblib')
dump(vectorizer, 'models/vectorizer.joblib')

print("Model and vectorizer saved successfully!")
```
